```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Ā\Documents\python_class\dataset\indianpincodes.csv")
```

```python
print(df.head())
```

```python
south_states = [
    "Andhra Pradesh",
    "Telangana",
    "Karnataka",
    "Kerala",
    "Tamilnadu"
]

south_india = df[df["Circle"].isin(south_states)].copy()
print(south_india.head())
```

```python
print("Total records in South India:", len(south_india))
```

```python
print("Unique pincodes in South India:", south_india["Pincode"].nunique())
```

```python
circle_count = south_india["Circle"].value_counts()

print("\nNumber of pincodes in each Circle:")
print(circle_count)
```

```python
# Q6. DataFrame containing only Tirupati Division
tirupati = south_india[
    south_india["Division"] == "Tirupati"
].copy()

print("\nTirupati Division records:")
print(tirupati)
```

```python
print("Number of pincodes in Tirupati:", tirupati["Pincode"].nunique())

print("\nFirst 10 records:")
print(tirupati.head(10))
```

```python
#Q8
top_10_divisions = df["Division"].value_counts().head(10)

print("\nTop 10 Divisions with highest number of pincodes:")
print(top_10_divisions)
```

```python
# Q9. Missing values in South India
missing = south_india.isna().sum().sort_values(
    ascending=False
)

print("\nMissing values in each column:")
print(missing)

print("\nColumn with the most missing values:")
print(missing.index[0])

print("Number of missing values:")
print(missing.iloc[0])
```

```python
# Q10. Pincodes in each Region within Andhra Pradesh
andhra_pradesh = south_india[
    south_india["Circle"] == "Andhra Pradesh"
].copy()

region_count = andhra_pradesh["Region"].value_counts()

print(
    "\nNumber of pincodes in each Region within Andhra Pradesh:"
)
print(region_count)
```

```python
# Q11. Save South India subset
south_india.to_csv(
    "south_india_pincodes.csv",
    index=False
)

print("\nSouth India dataset saved successfully!")
```
