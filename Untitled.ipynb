```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import folium
```

```python
pois = gpd.read_file("C:/Users/Ā/Documents/python_class/dataset/delhi_pois.geojson")
wards = gpd.read_file("C:/Users/Ā/Documents/python_class/dataset/wards_delhi_c.geojson")
```

```python
print("Number of POIs:", len(pois))
print("CRS:", pois.crs)
print("Columns:", pois.columns.tolist())

display(pois.head())
```

```python
pois["PoI_Type"].value_counts().head(20)
```

```python
pois_utm = pois.to_crs(epsg=32643)
wards_utm = wards.to_crs(epsg=32643)

print("POI CRS:", pois_utm.crs)
print("Ward CRS:", wards_utm.crs)
```

```python
health = pois_utm[
    pois_utm["PoI_Type"].str.contains(
        "hospital|clinic|health",
        case=False,
        na=False
    )
].copy()

print("Health-related POIs:", len(health))

display(
    health[["PoI_Type", "geometry"]].head()
)
```

```python
fig, ax = plt.subplots(figsize=(9, 8))

wards_utm.boundary.plot(
    ax=ax,
    color="grey",
    linewidth=0.4,
    zorder=1
)

health.plot(
    ax=ax,
    color="red",
    markersize=12,
    zorder=2
)

ax.set_title("Selected Health-related POIs in Delhi")
ax.set_axis_off()

plt.show()
```

```python
health_buffer = health.copy()

health_buffer["geometry"] = health_buffer.geometry.buffer(500)

display(
    health_buffer[["PoI_Type", "geometry"]].head()
)
```

```python
print("Original geometry:")
print(health.geom_type.value_counts())

print("\nBuffered geometry:")
print(health_buffer.geom_type.value_counts())
```

```python
fig, ax = plt.subplots(figsize=(10, 9))

wards_utm.boundary.plot(
    ax=ax,
    color="grey",
    linewidth=0.4,
    zorder=1
)

health_buffer.plot(
    ax=ax,
    alpha=0.30,
    edgecolor="black",
    linewidth=0.3,
    zorder=2
)

health.plot(
    ax=ax,
    color="red",
    markersize=10,
    zorder=3
)

ax.set_title("500 m Buffers Around Health-related POIs")
ax.set_axis_off()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 9))

wards_utm.boundary.plot(
    ax=ax,
    color="grey",
    linewidth=0.4,
    zorder=1
)

health_buffer.plot(
    ax=ax,
    alpha=0.30,
    edgecolor="black",
    linewidth=0.3,
    zorder=2
)

ax.set_title("500 m Buffers Around Health-related POIs")
ax.set_axis_off()

plt.show()
```

```python
health_coverage = health_buffer.dissolve()

fig, ax = plt.subplots(figsize=(10, 9))

wards_utm.boundary.plot(
    ax=ax,
    color="grey",
    linewidth=0.4,
    zorder=1
)

health_coverage.plot(
    ax=ax,
    alpha=0.35,
    edgecolor="black",
    zorder=2
)

health.plot(
    ax=ax,
    color="red",
    markersize=8,
    zorder=3
)

ax.set_title("Combined 500 m Health POI Buffer Coverage")
ax.set_axis_off()

plt.show()
```
