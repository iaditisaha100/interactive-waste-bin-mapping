```python
def dms_to_dd(degrees, minutes, seconds, hemisphere):
    decimal_degrees = degrees + minutes / 60 + seconds / 3600

    if hemisphere.upper() in {"S", "W"}:
        decimal_degrees = -decimal_degrees

    return decimal_degrees
```

```python
def valid_coordinate(longitude, latitude):
    valid_lon = -180 <= longitude <= 180
    valid_lat = -90 <= latitude <= 90

    return valid_lon and valid_lat
```

```python
latitude = dms_to_dd(13, 38, 23.6, "N")
longitude = dms_to_dd(79, 25, 39.2, "E")

tirupati_coordinate = (longitude, latitude)

print(f"Latitude: {latitude:.6f}")
print(f"Longitude: {longitude:.6f}")
print("Coordinate (longitude, latitude):", tirupati_coordinate)
print("Valid coordinate:", valid_coordinate(longitude, latitude))
```
