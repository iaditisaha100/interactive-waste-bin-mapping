# LECTURE 9 — From Python Collections to Pandas

The central question is:

> We already know how to store a few observations using lists and dictionaries. What happens when we want a proper table containing many observations and variables?

# 1. Why pandas?

Suppose we have information about three wards:

```python
ward_names = ["Ward 1", "Ward 2", "Ward 3"]
population = [12500, 18300, 9400]
area_km2 = [2.4, 3.1, 1.8]
```

These lists are useful, but the relationship between them is only something **we** understand.

A table makes that relationship explicit:

| Ward | Population | Area (km²) |
|---|---:|---:|
| Ward 1 | 12500 | 2.4 |
| Ward 2 | 18300 | 3.1 |
| Ward 3 | 9400 | 1.8 |

This is the kind of structure pandas calls a **DataFrame**.

# 2. Series — one labelled dimension

A **Series** is a one-dimensional labelled data structure.

A simple way to think about it is:

> **A Series is similar to one column of a table.**

````python
population = pd.Series([12500, 18300, 9400])
population
````

A Series has:

- values
- an index
- an optional name

````python
population = pd.Series(
    [12500, 18300, 9400],
    index=["W1", "W2", "W3"],
    name="Population"
)
population
````

````python
print(type(population))
print(population.index)
print(population.name)
````

## 2.1 Creating a Series from a dictionary

The dictionary keys become the Series index.

````python
park_distance_data = {
    "W1": 320,
    "W2": 780,
    "W3": 450
}

park_distance = pd.Series(
    park_distance_data,
    name="Park distance (m)"
)

park_distance
````

Connection to what we already know:

```text
Python dictionary
        ↓
key : value
        ↓
pandas Series
index : value
```

# 3. DataFrame — a labelled table

A **DataFrame** is a two-dimensional labelled table.

It contains:

- rows
- columns
- an index
- columns that can have different data types

Conceptually, a DataFrame can be thought of as a **collection of Series sharing a row index**.

## 3.1 Creating a DataFrame from a dictionary of lists

The dictionary keys become the column names.

````python
data = {
    "Ward": ["Ward 1", "Ward 2", "Ward 3"],
    "Population": [12500, 18300, 9400],
    "Area_km2": [2.4, 3.1, 1.8],
    "Land_use": ["Residential", "Mixed", "Residential"]
}

wards = pd.DataFrame(data)
wards
````

Notice:

```text
dictionary key      → column name
list                → column values
position in lists   → row
```

All lists should normally have the same length.

## 3.2 Creating a DataFrame from a list of lists

We can also provide rows directly and specify column names.

````python
data_rows = [
    ["Ward 1", 12500, 2.4, "Residential"],
    ["Ward 2", 18300, 3.1, "Mixed"],
    ["Ward 3", 9400, 1.8, "Residential"]
]

wards_from_rows = pd.DataFrame(
    data_rows,
    columns=["Ward", "Population", "Area_km2", "Land_use"]
)

wards_from_rows
````

The dictionary-of-lists version is often easier to read because the meaning of each list is explicitly labelled.

# 4. Creating a DataFrame from Series

This is useful because it reveals the relationship between **Series** and **DataFrame**.

Create several Series:

````python
ward = pd.Series(
    ["Ward 1", "Ward 2", "Ward 3"],
    index=["W1", "W2", "W3"],
    name="Ward"
)

population = pd.Series(
    [12500, 18300, 9400],
    index=["W1", "W2", "W3"],
    name="Population"
)

area = pd.Series(
    [2.4, 3.1, 1.8],
    index=["W1", "W2", "W3"],
    name="Area_km2"
)

land_use = pd.Series(
    ["Residential", "Mixed", "Residential"],
    index=["W1", "W2", "W3"],
    name="Land_use"
)
````

````python
ward_data = pd.DataFrame({
    "Ward": ward,
    "Population": population,
    "Area_km2": area,
    "Land_use": land_use
})

ward_data
````

Each Series became a column.

Because the Series share the same index (`W1`, `W2`, `W3`), pandas aligns them by index.

This idea of **alignment by labels** is one of the useful features of pandas.

# 11. Conditional filtering

Earlier we wrote logic such as:

```python
if country == "India":
```

With pandas, we can ask essentially the same question for **every row at once**.

````python
cities["country"] == "India"
````

The result is a Series of Boolean values:

```text
True
False
False
True
...
```

This is called a **Boolean mask**.

````python
india_mask = cities["country"] == "India"
type(india_mask)
````

````python
india_cities = cities[india_mask]
india_cities.head(10)
````

````python
cities[cities["country"] == "India"].head(10)
````

## 11.1 Numerical conditions

````python
cities[cities["population"] > 1_000_000].head(10)
````

Python allows underscores inside large numbers for readability:

```python
1_000_000
```

is the same value as `1000000`.

## 11.2 Combining conditions

For:

- country = India
- population > 1 million

use `&` for **AND**.

````python
large_indian_cities = cities[
    (cities["country"] == "India") &
    (cities["population"] > 1_000_000)
]

large_indian_cities[["city", "country", "population"]].head(20)
````

When combining pandas conditions, put each condition in parentheses.

Use:

```text
&   AND
|   OR
~   NOT
```

## 11.3 Spatial filtering using coordinates

Because the dataset contains latitude and longitude, we can filter by coordinate values.

For example:

- country = India
- latitude between 10° and 15° N

````python
india_lat_band = cities[
    (cities["country"] == "India") &
    (cities["lat"] >= 10) &
    (cities["lat"] <= 15)
]

india_lat_band[
    ["city", "admin_name", "lat", "lng", "population"]
].head(20)
````

This is a simple form of **spatial filtering** using coordinates.

Later, GeoPandas will let us work with actual geometry.

# LECTURE 11 — Transforming and Summarising Urban Data

We can now create, read, inspect, select, and filter data.

# 12. Creating new columns

````python
cities["population_millions"] = cities["population"] / 1_000_000

cities[
    ["city", "country", "population", "population_millions"]
].head()
````

This is a **vectorised operation**.

Pandas applies the calculation to the entire column.

We do not need a `for` loop for every row.

## 12.1 Calculated urban variables

```python
wards["density"] = wards["population"] / wards["area_km2"]
```

```text
existing columns
```

````python
cities["country"].value_counts(normalize=True).head(10)
````

# 16. Grouping observations with `groupby()`

`value_counts()` asks:

> How many observations are in each category?

`groupby()` asks:

> For each group, calculate some quantity.

Conceptually:

```text
SPLIT INTO GROUPS
       ↓
CALCULATE WITHIN EACH GROUP
       ↓
COMBINE RESULTS
```

## 16.1 Mean recorded city population by country

````python
mean_population_by_country = (
    cities
    .groupby("country")["population"]
    .mean()
    .sort_values(ascending=False)
)

mean_population_by_country.head(15)
````

## 16.2 Indian cities grouped by administrative region

````python
india = cities[cities["country"] == "India"].copy()
````

````python
india["admin_name"].value_counts().head(15)
````

````python
india.groupby("admin_name")["population"].mean().sort_values(
    ascending=False
).head(15)
````

The grouping variable defines **which observations belong together**.

The selected numerical column defines **what we summarise**.

The final method (`mean()`, `sum()`, `count()`, etc.) defines **how we summarise it**.
