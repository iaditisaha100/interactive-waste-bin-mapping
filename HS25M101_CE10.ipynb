```python
import pandas as pd
```

```python

# Index for five places
index = ["P1", "P2", "P3", "P4", "P5"]

# Place name Series
place = pd.Series(
    ["Ward 1", "Ward 2", "Ward 3", "Ward 4", "Ward 5"],
    index=index
)

# Population Series
population = pd.Series(
    [12000, 18500, 15000, 22000, 16500],
    index=index
)

# Area in square kilometres Series
area = pd.Series(
    [2.5, 3.2, 2.8, 4.1, 3.0],
    index=index
)

# Dominant land-use category Series
land_use = pd.Series(
    ["Residential", "Commercial", "Mixed", "Residential", "Industrial"],
    index=index
)
# Walking distance to nearest bus stop Series
bus_stop_distance = pd.Series(
    [250, 450, 700, 300, 850],
    index=index
)

# Combine all Series into one DataFrame
urban_places = pd.DataFrame({
    "Place": place,
    "Population": population,
    "Area_sq_km": area,
    "Land_use": land_use,
    "Bus_stop_distance_m": bus_stop_distance
})

# 1. Display the DataFrame
print("Urban Places:")
print(urban_places)

# 2. Print the shape
print("\nShape:")
print(urban_places.shape)

# 3. Print column names
print("\nColumn names:")
print(urban_places.columns)

# 4. Select only Place and Population
print("\nPlace and Population:")
print(urban_places[["Place", "Population"]])

# 5. Select only Bus_stop_distance_m
print("\nBus stop distance:")
print(urban_places["Bus_stop_distance_m"])

# Check whether the result is a Series or DataFrame
print("\nType:")
print(type(urban_places["Bus_stop_distance_m"]))
```
