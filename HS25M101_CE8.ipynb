```python
## defining  my park accessibility function
def classify_park_access(distance_m):
    if distance_m <= 300:
        return "Very good"
    elif distance_m <= 600:
        return "Good"
    elif distance_m <= 1000:
        return "Limited"
    else:
        return "Poor"

# classifying  the accessibility levels in the dictionary
ward_park_distance = {
    "Ward 1": 220,
    "Ward 2": 540,
    "Ward 3": 760,
    "Ward 4": 1350
}

for ward, distance in ward_park_distance.items():
    access = classify_park_access(distance)
    print(f"{ward}: {distance} m → {access}")
```
