```python
# ============================================================
# EXERCISE 16 — INTERACTIVE WASTE-BIN MAP USING FOLIUM
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------
import geopandas as gpd
import folium

```

```python

## 2. Load the Waste-Bin GeoPackage

data_path = r"C:\Users\Ā\Documents\python_class\HS25M101\HS25M101.gpkg"
dustbins = gpd.read_file(
    data_path,
    layer="notes_point"
)
```

```python

# 2. Inspect CRS
print(dustbins.crs)

# 3. Convert to EPSG:4326
dustbins = dustbins.to_crs(epsg=4326)

# Convert timestamp to string for Folium tooltip
dustbins["timestamp"] = dustbins["timestamp"].astype(str)

# 4. Find the centre of the study area
# 4. Calculate map centre
center = [
    dustbins.geometry.y.mean(),
    dustbins.geometry.x.mean()
]
```

```python
# 5. Create the Folium map
m = folium.Map(
    location=center,
    zoom_start=16,
    tiles=None
)
# 6. Add Esri World Imagery basemap
folium.TileLayer(
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}",
    attr="Esri World Imagery",
    name="Satellite"
).add_to(m)
```

```python
# 7. Add waste-bin points using GeoJson and CircleMarker
folium.GeoJson(
    dustbins,
    name="Waste Bins",
    tooltip=folium.GeoJsonTooltip(
        fields=["title", "note", "timestamp"],
        aliases=["Location", "Observation", "Recorded At"],
        localize=True
    ),
    marker=folium.CircleMarker(
        radius=6,
        fill=True,
        fill_opacity=0.8,
        weight=1
    )
).add_to(m)

# 8. Add layer control
folium.LayerControl().add_to(m)

# 9. Display the interactive map
m
```

```python

# 10. Save the map as an HTML file
m.save("HS25M101_Exercise16_WasteBin_Map.html")
```
