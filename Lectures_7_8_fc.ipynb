# 1. Making decisions with `if`, `elif`, and `else`

A large part of data analysis involves asking:

> **Does this observation satisfy a condition?**

Examples from urban and spatial analysis:

- Is a bus stop within a given walking distance?
- Is a population-density value above a threshold?
- Is a latitude value valid?
- Is a survey observation complete?
- Should a ward be flagged for further investigation?

Python uses **conditional statements** to make these decisions.

## 1.1 The basic idea

The general structure is:

```python
if condition:
    statement
elif another_condition:
    statement
else:
    statement
```

Python evaluates the conditions from **top to bottom**.

- `if` checks the first condition.
- `elif` means **else if** and checks another condition if the earlier one was false.
- `else` handles everything that did not satisfy the earlier conditions.

### Indentation matters

The indented lines belong to the condition.

```python
if condition:
    this_line_runs_if_true
```

Unlike many languages, indentation in Python is not only for appearance — it is part of the syntax.

## 1.2 First example: distance to a bus stop

Suppose we have measured the walking distance from a residential location to the nearest bus stop.

For demonstration, we will use these **illustrative categories**:

- up to 400 m → `"High access"`
- more than 400 m and up to 800 m → `"Moderate access"`
- more than 800 m → `"Low access"`

These values are being used only to practise conditional logic.

````python
distance = 620  # metres

if distance <= 400:
    access = "High access"
elif distance <= 800:
    access = "Moderate access"
else:
    access = "Low access"

print(access)
````

````python
distance = 400 

if distance <=200:
    access = "Very good"
elif distance <=400:
    access = "Good"
elif distance <=800:
    access = "Decent" 
else:
    access = "Bad" 
    
print(access)
````

### Why does the second condition only say `<= 800`?

Because Python has already tested:

```python
distance <= 400
```

If Python reaches the `elif`, it already knows that the distance is **greater than 400**.

Therefore:

```python
elif distance <= 800:
```

effectively means:

> greater than 400 **and** up to 800.

The **order of conditions matters**.

````python
# Try changing this value to 250, 600, and 1200
distance= 250

if distance <= 400:
    print("High access")
elif distance <= 800:
    print("Moderate access")
else:
    print("Low access")
````

## 2.3 `.split()` turns one string into a list

This is extremely useful when one field contains several pieces of information.

Suppose a field survey record arrives as:

```text
Ward 12 | Residential | Waterlogging
```

We can split the string wherever `|` occurs.

````python
record = "Ward 12 | Residential | Waterlogging"

parts = record.split("|")

print(parts)
print(type(parts))
````

The resulting values still contain some spaces.  
We can clean the individual pieces using `.strip()`.

````python
record = "Ward 12 | Residential | Waterlogging"
parts = record.split("|")

ward = parts[0].strip()
land_use = parts[1].strip()
observation = parts[2].strip()

print(ward)
print(land_use)
print(observation)
````

## 2.4 Parsing a coordinate copied as text

A coordinate copied from a source may initially arrive as a string:

```text
"13.6288, 79.4192"
```

Python does not yet see two numerical coordinates. It sees **one string**.

We can:

1. split the string at the comma
2. remove unnecessary spaces
3. convert the two pieces to floats
4. store them in the coordinate order we want

Here, the source text is in `(latitude, longitude)` order, while our spatial tuple will use `(longitude, latitude)`.

````python
coordinate_text = "13.6288, 79.4192"

parts = coordinate_text.split(",")

latitude = float(parts[0].strip())
longitude = float(parts[1].strip())

coordinate = (longitude, latitude)

print("Latitude:", latitude)
print("Longitude:", longitude)
print("Spatial tuple (longitude, latitude):", coordinate)
````

# 4. Why do we need loops?

So far, we have mostly processed **one value at a time**.

But spatial datasets contain many observations:

- many bus stops

A **loop** allows Python to repeat an operation.

# 5. The `for` loop

Use a `for` loop when you want to iterate through the items in a collection.

General structure:

```python
for item in collection:
    do_something
```

In plain English it is something like: 

> **For each item in this collection, perform these statements.**

## 5.1 Looping through a route

Our route is stored as a list.

Instead of accessing every stop manually, we can ask Python to visit every element.

````python
route = [
    "IIT Tirupati",
    "Yerpedu",
    "Renigunta",
    "Tirupati Bus Stand"
]

for stop in route:
    print(stop)
````

The variable `stop` is temporary.

During each iteration:

```text
1st iteration → stop = "IIT Tirupati"
2nd iteration → stop = "Yerpedu"
3rd iteration → stop = "Renigunta"
4th iteration → stop = "Tirupati Bus Stand"
```

Python stops automatically when it reaches the end of the list.

````python
for stop in route:
    print(f"Bus reaches: {stop}")
````

## 5.2 Using `range()` when we want repeated numbers

Sometimes we want a loop to repeat a known number of times rather than move through an existing list.

`range()` generates a sequence of integers.

For example, suppose we want to create labels for five field-survey locations:

````python
for number in range(1, 6):
    print(f"Survey Location {number}")
````

`range(1, 6)` produces:

```text
1, 2, 3, 4, 5
```

The ending value `6` is **not included**.

This follows the same stop-before-the-end idea that you saw earlier in Python slicing.

## 5.3 Looping through a dictionary

Suppose we know the walking distance to the nearest park for four wards:

````python
park_distance = {
    "Ward 1": 320,
    "Ward 2": 780,
    "Ward 3": 450,
    "Ward 4": 1100
}

for ward, distance in park_distance.items():
    print(f"{ward}: {distance} m")
````

`.items()` gives us the **key and value together**:

```text
ward     → dictionary key
distance → dictionary value
```

This is especially useful for data stored as:

```python
{spatial_unit: value}
```

## 5.4 Combining a loop with an `if` condition

This is where programming becomes much more useful.

We can inspect every ward and identify only those that satisfy a condition.

````python
park_distance = {
    "Ward 1": 320,
    "Ward 2": 780,
    "Ward 3": 450,
    "Ward 4": 1100
}

for ward, distance in park_distance.items():
    if distance > 500:
        print(f"{ward} is more than 500 m from the nearest park.")
````

## 5.5 Building a new list using a loop

Instead of only printing the matching wards, we can **store them** for later analysis.

````python
underserved_wards = []

for ward, distance in park_distance.items():
    if distance > 500:
        underserved_wards.append(ward)

print(underserved_wards)
````

Notice how several concepts are now working together:

```text
dictionary
    ↓
for loop
    ↓
if condition
    ↓
append()
    ↓
new list
```

This pattern appears repeatedly in data analysis.

## 5.6 Accumulators: counting and adding inside a loop

A variable can accumulate information while the loop runs.

For example, we can count how many wards satisfy a condition without using `len()` on a filtered list.

````python
count = 0

for ward, distance in park_distance.items():
    if distance > 500:
        count += 1

print(f"Number of wards more than 500 m from a park: {count}")
````

````python
tree_counts = {
    "Ward 1": 125,
    "Ward 2": 80,
    "Ward 3": 210,
    "Ward 4": 95
}

total_trees = 0

for ward, trees in tree_counts.items():
    total_trees += trees

print(f"Total recorded trees:{total_trees}")
````

# 6. Controlling a loop: `break` and `continue`

Sometimes we do not want every iteration to behave in exactly the same way.

Two useful commands are:

- `break` → **stop the entire loop immediately**
- `continue` → **skip the current item and move to the next iteration**

## 6.1 `break`: stop when we have found what we need

Imagine scanning an ordered route until we reach Renigunta.

````python
route = [
    "IIT Tirupati",
    "Yerpedu",
    "Renigunta",
    "Tirupati Bus Stand"
]

for stop in route:
    print("Checking:", stop)

    if stop == "Renigunta":
        print("Renigunta found. Stop searching.")
        break
````

Once `break` runs, later stops are not checked.

Use `break` when:

> **the purpose of the loop has already been achieved.**

## 6.2 `continue`: skip unusable observations

Real datasets often contain missing or invalid values.

Suppose `None` represents a missing pedestrian count:

````python
pedestrian_counts = [120, 95, None, 160, None, 140]

for count in pedestrian_counts:
    if count is None:
        continue

    print("Valid count:", count)
````

````python
fruits = ["apple", "banana", "mango", "orange"]

for fruit in fruits:
    if fruit == "banana":
        continue   # skip this one
    print("I like", fruit)
````

Here, `continue` does **not** stop the loop.

It means:

> Ignore this observation and continue with the next one.

# 7. The `while` loop

A `for` loop is useful when we have a collection and want to process its items.

A `while` loop is useful when we want to continue repeating something **while a condition remains true**.

General form:

```python
while condition:
    statements
```

A useful way to think about the difference is:

| `for` | `while` |
|---|---|
| repeat for each item | repeat while a condition is true |
| number of iterations is often known from the collection | number of iterations may not be known in advance |

## 7.1 Example: validating latitude

A latitude must lie between `-90` and `+90`.

If a field-data operator enters an invalid value, we can keep asking until a valid value is supplied.

> Run the following cell interactively.

````python
latitude = float(input("Enter latitude: "))

while latitude < -90 or latitude > 90:
    print("Invalid latitude. Please try again.")
    latitude = float(input("Enter latitude again: "))

print(f"Accepted latitude: {latitude}")
````

### Be careful with infinite loops

A `while` loop needs some way for its condition eventually to become `False`.

In the latitude example, the value of `latitude` is updated inside the loop. If the value never changes, the loop may continue forever.

A safe pattern is:

```python
number = 1

while number <= 5:
    print(number)
    number += 1
```

The important idea is:

```text
while condition:
    do something
    update something that can eventually make the condition False
```

# 8. Functions: reusable pieces of analysis

A **function** is a reusable block of code designed to perform a specific task.

Why are functions important in spatial analysis?

Suppose we have a classification rule that must be applied to:

- 5 wards
- 500 buildings
- 20,000 survey observations

We should not rewrite the same `if` statement thousands of times.

Instead, we can define the rule **once** and reuse it.

General structure:

```python
def function_name(parameters):
    statements
    return result
```

## 8.1 A first spatial function

The transit-access rule from Lecture 7 can be turned into a reusable function.


````python
def classify_transit_access(distance_m):
    """Classify transit access using illustrative walking-distance thresholds."""

    if distance_m <= 400:
        return "High access"
    elif distance_m <= 800:
        return "Moderate access"
    else:
        return "Low access"
````

The text inside triple quotes is a **docstring**. It briefly explains what the function does.

Because we documented the function, `help()` can show that information:

````python
help(classify_transit_access)
````

````python
print(classify_transit_access(250))
print(classify_transit_access(650))
print(classify_transit_access(1200))
````

### What happens when we call the function?

```python
classify_transit_access(650)
```

Here:

- `650` is passed into the function
- inside the function, `distance_m` becomes `650`
- Python evaluates the conditions
- `return` sends `"Moderate access"` back to us

The function does not care whether the value belongs to Ward 1, Ward 2, or any other place.  
It performs the same rule whenever we give it a distance.

## 8.2 `print()` and `return` are different

This distinction is important.

`print()` **displays** something.

`return` **gives a value back** so that the rest of the program can use it.

````python
def density_value(population, area_sq_km):
    density = population / area_sq_km
    return density

ward_density = density_value(18450, 3.2)

print(ward_density)
print(f"Density: {ward_density:.1f} persons per sq. km")
````

Because the function **returns** the density, we can:

- store it
- print it
- compare it
- use it in another calculation
- place it inside a larger workflow

## 8.3 Functions can have several parameters

A coordinate needs both longitude and latitude.

We can create a function that checks whether both values are valid.

````python
def valid_coordinate(longitude, latitude):
    valid_lon = -180 <= longitude <= 180
    valid_lat = -90 <= latitude <= 90

    return valid_lon and valid_lat
````

````python
print(valid_coordinate(79.419, 13.628))
print(valid_coordinate(250, 13.628))
````

## 8.4 Functions become powerful when combined with loops

Now we can apply the **same classification rule to every ward**.

````python
bus_stop_distance = {
    "Ward A": 250,
    "Ward B": 610,
    "Ward C": 980,
    "Ward D": 390
}

for ward, distance in bus_stop_distance.items():
    access = classify_transit_access(distance)
    print(f"{ward}: {distance} m → {access}")
````

This is an important programming pattern:

```text
DATA
 ↓

 ↓
FUNCTION applies the same rule

RESULT for every observation
```

# 9. Spatial application of a function: DMS to decimal degrees

Geographic coordinates may be written in **degrees, minutes, and seconds (DMS)**:

```text
13° 37′ 44.634″ N
```

or in **decimal degrees (DD)**:

```text
13.629065°
```

The conversion is:

\[
DD = Degrees + \frac{Minutes}{60} + \frac{Seconds}{3600}
\]

For southern and western hemispheres, the decimal-degree value is negative.

In our geographic-coordinate examples, we store points as:

```python
(longitude, latitude)
```

which corresponds to the common spatial-programming order:

```text
(x, y)
```

So after converting latitude and longitude to decimal degrees, we can place them directly into a coordinate tuple.

This is a good example of a calculation that we should place inside a reusable function.

````python
def dms_to_dd(degrees, minutes, seconds, hemisphere):
    decimal_degrees = degrees + minutes / 60 + seconds / 3600

    if hemisphere.upper() in {"S", "W"}:
        decimal_degrees = -decimal_degrees

    return decimal_degrees
````

We can now reuse the function for both latitude and longitude.

````python
latitude = dms_to_dd(13, 37, 44.6340, "N")
longitude = dms_to_dd(79, 25, 28.0056, "E")

tirupati_coordinate = (longitude, latitude)

print(f"Latitude: {latitude:.6f}")
print(f"Longitude: {longitude:.6f}")
print("Coordinate (longitude, latitude):", tirupati_coordinate)
````

Notice how this example combines several ideas from the course:

- numbers
- strings (`"N"`, `"E"`)
- a set (`{"S", "W"}`)
- an `if` condition
- a function
- `return`
- a coordinate tuple
- f-string formatting

# 10. Optional extension: recursion

A **recursive function** is a function that calls itself.

Recursion is useful for problems that can be broken into smaller versions of the same problem.

Every recursive function needs:

1. a **base case** — when to stop
2. a **recursive case** — when the function calls itself again

For beginners, loops are usually easier for processing spatial records.  
Recursion becomes more useful later for hierarchical or network-like structures.

## A simple route example

The following function prints the first stop and then calls itself with the remaining route.

The base case occurs when no stops remain.

````python
def visit_route(stops):
    if len(stops) == 0:
        return

    print("Visiting:", stops[0])
    visit_route(stops[1:])


route = [
    "IIT Tirupati",
    "Yerpedu",
    "Renigunta",
    "Tirupati Bus Stand"
]

visit_route(route)
````

For this particular task, a `for` loop would be simpler.

The purpose of the example is only to understand the idea:

```text
function
  ↓
solves one small part
  ↓
calls itself for what remains
  ↓
stops at the base case
```
