```python
print("\nColumn names:")
print(cities.columns)
```

```python
selected_cities = cities[[
    "city",
    "country",
    "admin_name",
    "lat",
    "lng",
    "population"
]]
```

```python
print("\nSelected columns:")
print(selected_cities)
```

```python
# 6. Create a DataFrame containing only cities in India
indian_cities = cities[cities["country"] == "India"]
```

```python
print("\nIndian cities:")
print(indian_cities)
```

```python
# 7. Indian cities with population greater than 1,000,000
large_indian_cities = indian_cities[
    indian_cities["population"] > 1000000
]
```

```python
print("\nIndian cities with population greater than 1,000,000:")
print(large_indian_cities)
```

```python
# 8. Indian cities whose latitude lies between 12° and 14° N
latitude_cities = indian_cities[
    (indian_cities["lat"] >= 12) &
    (indian_cities["lat"] <= 14)
]

print("\nIndian cities between 12° and 14° N:")
print(latitude_cities)
```

```python
# 9. Display only city, administrative region, latitude, longitude and population
final_result = latitude_cities[[
    "city",
    "admin_name",
    "lat",
    "lng",
    "population"
]]

print("\nFinal result:")
print(final_result)
```
