```python
import geopandas as gpd

bengaluru_gdf = gpd.read_file("C:/Users/Ā/Documents/python_class/dataset/Bengaluru_boundary.geojson")

bengaluru_gdf
```

```python
# Check the data
print("CRS:", bengaluru_gdf.crs)
print("Shape:", bengaluru_gdf.shape)

bengaluru_gdf.head()
```

```python
bengaluru = geemap.geopandas_to_ee(bengaluru_gdf)

print("Bengaluru boundary converted to Earth Engine format.")
```

```python
# Load the monthly UTCI dataset
utci_collection = ee.ImageCollection(
    "projects/sat-io/open-datasets/gloutci-m"
)

# Check the total number of images
print("Total images:", utci_collection.size().getInfo())
```

```python
utci_summer = (
    utci_collection
    .filterBounds(bengaluru)
    .filterDate("2020-03-01", "2020-06-01")
)

print("Images for Bengaluru, March–May 2020:",
      utci_summer.size().getInfo())
```

```python
# Get the first UTCI image
first_image = ee.Image(utci_summer.first())

# Check band names
print("Band names:", first_image.bandNames().getInfo())

# Check image properties
print("Image properties:", first_image.propertyNames().getInfo())
```

```python
# Calculate the mean UTCI for March–May 2020
utci_mean = (
    utci_summer
    .mean()
    .divide(100)
    .clip(bengaluru)
)

print("Seasonal mean UTCI calculated.")
```

```python
# Create an interactive map
Map = geemap.Map()

Map.centerObject(bengaluru, 10)

# Add Bengaluru boundary
Map.addLayer(
    bengaluru,
    {},
    "Bengaluru Boundary"
)

# Add mean UTCI layer
utci_vis = {
    "min": 20,
    "max": 40,
    "palette": ["blue", "cyan", "yellow", "orange", "red"]
}

Map.addLayer(
    utci_mean,
    utci_vis,
    "Mean UTCI (March–May 2020)"
)

Map
```

```python
# Identify areas with UTCI >= 32°C
high_utci = utci_mean.gte(32)

# Clip the result to Bengaluru
high_utci = high_utci.clip(bengaluru)

print("High-UTCI layer created successfully.")
```

```python
Map.addLayer(
    high_utci.selfMask(),
    {"palette": ["red"]},
    "High UTCI (>=32°C)"
)

Map
```

```python
# Calculate minimum, mean and maximum UTCI
utci_stats = utci_mean.reduceRegion(
    reducer=(
        ee.Reducer.min()
        .combine(
            ee.Reducer.mean(),
            sharedInputs=True
        )
        .combine(
            ee.Reducer.max(),
            sharedInputs=True
        )
    ),
    geometry=bengaluru,
    scale=1000,
    maxPixels=1e9
)

print(utci_stats.getInfo())
```

```python
# Count pixels below and above the UTCI threshold
utci_categories = (
    high_utci
    .rename("high_utci")
    .reduceRegion(
        reducer=ee.Reducer.frequencyHistogram(),
        geometry=bengaluru,
        scale=1000,
        maxPixels=1e9
    )
)

print(utci_categories.getInfo())
```

```python
# Create an image with pixel area in square kilometres
pixel_area = ee.Image.pixelArea().divide(1e6)

# Calculate total Bengaluru area
total_area = pixel_area.reduceRegion(
    reducer=ee.Reducer.sum(),
    geometry=bengaluru,
    scale=1000,
    maxPixels=1e9
)

# Calculate area where UTCI >= 32°C
high_area = pixel_area.updateMask(high_utci).reduceRegion(
    reducer=ee.Reducer.sum(),
    geometry=bengaluru,
    scale=1000,
    maxPixels=1e9
)

print("Total area (sq. km):", total_area.getInfo())
print("High UTCI area (sq. km):", high_area.getInfo())
```

```python
# Load WorldPop population data for 2020
population = (
    ee.ImageCollection("WorldPop/GP/100m/pop")
    .filterBounds(bengaluru)
    .filterDate("2020-01-01", "2021-01-01")
    .mosaic()
    .select("population")
    .clip(bengaluru)
)

print("WorldPop population data loaded successfully.")
```

```python
# Display population data
Map.addLayer(
    population,
    {
        "min": 0,
        "max": 1000,
        "palette": ["white", "yellow", "orange", "red"]
    },
    "WorldPop Population 2020"
)

Map
```

```python
# Calculate the total population in Bengaluru
total_population = ee.Number(
    population.reduceRegion(
        reducer=ee.Reducer.sum(),
        geometry=bengaluru,
        scale=100,
        maxPixels=1e9
    ).get("population")
)

print("Total population:", total_population.getInfo())
```

```python
# Align the high-UTCI mask with the population raster
high_utci_100m = high_utci.resample("nearest")

# Keep population values only in high-UTCI areas
population_high_utci = population.updateMask(high_utci_100m)

print("Population in high-UTCI areas calculated.")
```
