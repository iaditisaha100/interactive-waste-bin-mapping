```python
record = "  ward 7 | COMMERCIAL | waterlogging near junction  "
```

```python
# Remove unnecessary spaces
record = record.strip()
```

```python
# Split into three components
parts = record.split("|")
```

```python
# Clean each component
ward = parts[0].strip().title()
land_use = parts[1].strip().title()
observation = parts[2].strip().title()
```

```python
# Print using f-string
print(f"{ward} | Land use: {land_use} |")
print(f"Observation: {observation}")
```
