```python
name_person1 = input("hey person 1 , what is your name?:")
name_person2 = input("hey person 2 , what is your name?:")
```

```python
age_person1 = int(input("enter the age of person1:"))
age_person2 = int(input("enter the age of person2:"))
```

```python
comb_age= age_person1 +age_person2
```

```python
print (comb_age)
```

```python
# 1. Automatic placeholders
print("Meet {} ({} years old) and {} ({} years old)!".format(
    name_person1, age_person1, name_person2, age_person2
))
```

```python
# 2. Indexed placeholders
print("Meet {0} ({1} years old) and {2} ({3} years old)!".format(
    name_person1, age_person1, name_person2, age_person2
))
```

```python
# 3. Named placeholders
print("Meet {n1} ({a1} years old) and {n2} ({a2} years old)!".format(
    n1=name_person1, a1=age_person1, n2=name_person2, a2=age_person2
))
```

```python
# 4. f-string
print(f"Meet {name_person1} ({age_person1} years old) and {name_person2} ({age_person2} years old)!")
```

```python
# Combined age
print(f"Together they have a combined age of {comb_age} years.")
```
