```python
park_distance = {
    "Ward A": 280,
    "Ward B": 640,
    "Ward C": 920,
    "Ward D": 410,
    "Ward E": 1250
}
print (park_distance)
```

```python
priority_wards = []
count = 0 
for i, j in park_distance.items ():
    if j > 500 :
         priority_wards.append(i)
         count = count+1
    print(f"{i} has the nearest park at a distance of {j} m")
print(f"Total priority wards are: {priority_wards}")    
print(f"the Total number of priority wards are {count}")  
```
