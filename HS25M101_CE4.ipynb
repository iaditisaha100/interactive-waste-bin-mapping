```python
distance = float(input("Enter the walking distance to the nearest bus stop(in metres):"))
```

```python
if distance <= 400:
    access = "High access"
elif distance <=800:
    access = "Moderate access"
else:
    access = "Low access"
```

```python
print(f"The nearest bus stop is {distance} m away. Transit access:{access}.")
```
