```python
# 6. Create a DataFrame containing only cities in the United States
us_cities = cities[cities["country"] == "United States"]

print("United States cities:")
print(us_cities)
```

```python
# US cities with population greater than 2,000,000

large_us_cities = us_cities[
    us_cities["population"] > 2000000
]

print("US cities with population greater than 2,000,000:")
print(large_us_cities)
```

```python
print("Number of large US cities:", len(large_us_cities))
```

```python
cities["population_millions"] = cities["population"] / 1_000_000

cities[
    ["city", "country", "population", "population_millions"]
].head()
```
