```python
# ============================================================
# CLASS EXERCISE 12
# ============================================================

# Question:
# Using only POIs where category == "Education":
#
# 1. Create a new GeoDataFrame containing only Education POIs.
# 2. Spatially join these points to the Delhi wards.
# 3. Display poi_type, ward_id, and ward_name.
# 4. Count how many Education POIs did not match a ward.
#
# Question:
# Why was a spatial join necessary here rather than
# a normal Pandas merge?
```

```python
# ------------------------------------------------------------
# 8. SELECT ONLY EDUCATION POIs
# ------------------------------------------------------------

education = pois[
    pois["category"] == "Education"
].copy()
```

```python
# Display Education POIs

print("Number of Education POIs:", len(education))

display(education.head())

```

```python
# ------------------------------------------------------------
# 9. CHECK CRS BEFORE SPATIAL JOIN
# ------------------------------------------------------------

print("Education POI CRS:", education.crs)
print("Ward CRS:", wards.crs)
```

```python
# ------------------------------------------------------------
# 10. MAKE CRS SAME IF NECESSARY
# ------------------------------------------------------------

if education.crs != wards.crs:
    education = education.to_crs(wards.crs)


print("CRS after checking:")
print("Education:", education.crs)
print("Wards:", wards.crs)
```

```python
# ------------------------------------------------------------
# 11. SPATIAL JOIN
# ------------------------------------------------------------

education_joined = gpd.sjoin(
    education,
    wards[
        [
            "ward_id",
            "ward_name",
            "geometry"
        ]
    ],
    how="left",
    predicate="within"
)
```

```python
# ------------------------------------------------------------
# 12. REMOVE index_right
# ------------------------------------------------------------

education_joined = education_joined.drop(
    columns="index_right"
)

```

```python
# ------------------------------------------------------------
# 13. DISPLAY REQUIRED COLUMNS
# ------------------------------------------------------------

display(
    education_joined[
        [
            "poi_type",
            "ward_id",
            "ward_name"
        ]
    ]
)
```

```python
# ------------------------------------------------------------
# 14. COUNT EDUCATION POIs THAT DID NOT MATCH A WARD
# ------------------------------------------------------------

unmatched_education = (
    education_joined["ward_name"]
    .isna()
    .sum()
)


print(
    "Number of Education POIs that did not match a ward:",
    unmatched_education
)
```

```python
# ------------------------------------------------------------
# 15. OPTIONAL: DISPLAY UNMATCHED POIs
# ------------------------------------------------------------

unmatched = education_joined[
    education_joined["ward_name"].isna()
]


print("Unmatched Education POIs:")

display(
    unmatched[
        [
            "poi_type",
            "ward_id",
            "ward_name",
            "geometry"
        ]
    ]
)
```
