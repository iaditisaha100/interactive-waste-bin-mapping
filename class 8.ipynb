```python
def square_x(x):
    y = x**2
    return y
```

```python
square_x(2)
```

```python
def aditi(x,y):
    z = (x+y)**2
    return z
```

```python
aditi(2,3)
```

```python
## 
def factorial_n(x):
    y = 1 
    for i in range (1, x+1):
        y = y*i
    return y
```

```python
factorial_n(3)
```
